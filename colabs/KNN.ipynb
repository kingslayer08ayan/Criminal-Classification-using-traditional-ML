```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score

# Create a KNN Classifier
knn = KNeighborsClassifier(n_neighbors=3)  # weights = uniform || distance

# Train the model
knn.fit(X_train, y_train)

# Make predictions
y_pred = knn.predict(X_test)

# Evaluate the model
print(f"Accuracy: {accuracy_score(y_test, y_pred)}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

Cross Validation Testing

```python
from sklearn.model_selection import cross_val_score
# Test with both 'uniform' and 'distance' weights using cross-validation
for weight in ['uniform', 'distance']:
    knn = KNeighborsClassifier(n_neighbors=5, weights=weight)
    scores = cross_val_score(knn, X_train, y_train, cv=5, scoring='accuracy')
    print(f"Weight: {weight}, Cross-Validation Accuracy: {scores.mean():.4f}")
```
